```python
df = pd.DataFrame(data)
avg = df["personal_best_minutes"].mean()
df["personal_best_minutes"].fillna(avg,inplace=True)
```

```python
print("avg age is",df['age'].mean())
print("median:",df['age'].median())
print("mode is:",df['age'].mode())
print("standard deviation:",df['age'].std())
print("variance:",df['age'].var())
```

```python
print("avg previous_marathon_count	is",df['previous_marathon_count'].mean())
print("median:",df['previous_marathon_count'].median())
print("mode is:",df['previous_marathon_count'].mode())
print("standard deviation:",df['previous_marathon_count'].std())
print("variance:",df['previous_marathon_count'].var())
```

##Exploratory Data analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.hist(df['age'])
plt.xlabel("age")

plt.show()
```

```python
sns.countplot(x='gender', data=df)
plt.show()
```

```python
sns.scatterplot(data=df,x="weekly_mileage_km",y='personal_best_minutes')
plt.xlabel("weekly_mileage_km")
plt.ylabel("personal_best_minutes")
plt.title("weekly_mileage_km vs personal_best_minutes ")
plt.show()
```

```python
sns.barplot(data=df,x="gender",y='personal_best_minutes')
plt.xlabel("gender")
plt.ylabel("personal_best_minutes")
plt.title("gendervspersonal_best_minutes ")
plt.show()
```

```python
sns.lineplot(data=df,x="previous_marathon_count",y='personal_best_minutes')
plt.xlabel("previous_marathon_count")
plt.ylabel("personal_best_minutes")
plt.title("previous_marathon_count vs personal_best_minutes")
plt.show()
```

```python
gender_counts = df['gender'].value_counts()
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90)
plt.title("Gender Distribution")
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
sns.pairplot(df)
plt.show
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(20,15))

plt.subplot(3,3,1)
sns.histplot(df['age'],kde=True)

plt.subplot(3,3,2)
sns.countplot(data=df,x="gender")

plt.subplot(3,3,3)
sns.barplot(data=df,x="gender",y="personal_best_minutes")

plt.subplot(3,3,4)
sns.lineplot(data=df,x="previous_marathon_count",y="personal_best_minutes",color='green')
plt.title("previous_marathon_count vs personal_best_minutes")

plt.subplot(3,3,5)
gender_counts = df['gender'].value_counts()
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90)
plt.title("Gender Distribution")
plt.axis('equal')

plt.subplot(3,3,6)
corr=df.corr(numeric_only=True)
sns.heatmap(corr,annot=True)

plt.tight_layout()
plt.show()
```
